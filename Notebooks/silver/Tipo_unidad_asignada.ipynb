Cargar librerias

```python
from pyspark.sql.functions import current_timestamp
```

```python
path_base = 'abfss://silver@datapractica.dfs.core.windows.net' 
```

1. Extraer data de clientes

```python
uasignada_df = spark.read.format('delta').load(f'{path_base}/Vehiculos/')
```

2. Seleccionar tipo de viaje spartcab

```python
unidad_asignada_df = uasignada_df.select('Tipo_movil_Smartcab').distinct()
#crear timestamp
unidad_asignada_df = unidad_asignada_df.withColumn('timestamp', current_timestamp())
display(unidad_asignada_df)
```

3. Agregar la data como silver

```python
unidad_asignada_df.write.mode('overwrite') \
        .format('delta') \
        .save(f'{path_base}/UnidadAsignada/') 
```
